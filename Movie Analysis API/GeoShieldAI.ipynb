```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')  # Nano model — fast, lightweight

model.train(
    data='/content/crack_dataset/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    name='crack_detection_model'
)
```

```python
from ultralytics import YOLO
from PIL import Image
import cv2
import matplotlib.pyplot as plt

# Load best model
model = YOLO('runs/detect/crack_detection_model/weights/best.pt')

# Load one of your test images
img_path = '/content/crack_dataset/test/images/Gemini_Generated_Image_bnuk05bnuk05bnuk_png.rf.6e46370ebb9634891955b4993fa3dc20.jpg'  # Change to any test image

# Predict
results = model(img_path)

# Show result
result_img = results[0].plot()  # Draws bounding box/polygon
plt.figure(figsize=(10,8))
plt.imshow(result_img)
plt.axis('off')
plt.title("Crack Detected by Trained Model")
plt.show()
```
